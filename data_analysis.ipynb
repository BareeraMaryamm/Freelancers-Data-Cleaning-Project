```python
#Rename Columns

df.rename(columns = {"is_Active" : "Active_Status"}, inplace = True)
df.head()
```

```python
# Check Duplicates

df.duplicated().sum()
```

```python
# Null check

print(df.isnull().sum()) 
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">✨</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">NAME CLEANING & STANDARDIZATION</span>
</div>

```python
df["Name"] = df['Name'].str.replace('Mr.|Ms.|Mrs.|Dr.|Miss', '', regex=True).str.strip()

# Remove DDS, MD, PhD from last
df['Name'] = df['Name'].str.replace(' DDS| MD| PhD', '', regex=True)
df["Name"].head(10)
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">⚧️</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">GENDER CLEANING & NORMALIZATION</span>
</div>

```python
df["Gender"] = df["Gender"].replace({
    "f": "Female",
    "F": "Female", 
    "FEMALE": "Female",
    "female": "Female",
    "Female": "Female",
    "m": "Male",
    "M": "Male",
    "MALE": "Male",
    "male": "Male",
    "Male": "Male"
})
df["Gender"] 
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">🎂</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">AGE IMPUTATION & TYPE CONVERSION</span>
</div>

```python
df["Age"]= df["Age"].fillna(df["Age"].median())
df['Age'] = df['Age'].astype(int)
df["Age"]
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">💼</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">YEARS OF EXPERIENCE - IMPUTATION</span>
</div>

```python
df["Years_of_Experience"] = df["Years_of_Experience"].fillna(df["Years_of_Experience"].median())
df["Years_of_Experience"] = df["Years_of_Experience"].astype(int)
df["Years_of_Experience"]
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">💵</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">HOURLY RATE - CLEANING & CONVERSION</span>
</div>

```python
# It remove $ , USD , text and only numbers will shown

df['Hourly_Rate (USD)'] = df['Hourly_Rate (USD)'].astype(str).str.replace(r'[^0-9.]', '', regex=True)
df['Hourly_Rate (USD)'] = pd.to_numeric(df['Hourly_Rate (USD)'], errors='coerce').fillna(0).astype(int)

print(df['Hourly_Rate (USD)'].dtype) # int64
print(df['Hourly_Rate (USD)'].head(10))
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">⭐</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">RATING - MEDIAN IMPUTATION</span>
</div>

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].median())
df["Rating"]
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">✅</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">ACTIVE STATUS - BOOLEAN STANDARDIZATION</span>
</div>

```python
df['Active_Status'] = df['Active_Status'].astype(str).str.lower().map({
    '0':'False','1':'True','n':'False','y':'True',
    'no':'False','yes':'True','false':'False','true':'True'
})

df['Active_Status'] = df['Active_Status'].fillna(df['Active_Status'].mode()[0])

df['Active_Status']
```

<div style="background: linear-gradient(135deg, #fdf2f8 0%, #fce7f3 25%, #dbeafe 100%); padding:11px 20px; border-radius:10px; font-family:'Segoe UI', sans-serif; border:1px solid #fbcfe8; display:inline-flex; align-items:center; gap:10px;">
  <span style="font-size:18px;">📊</span>
  <span style="font-size:13px; font-weight:800; color:#831843; letter-spacing:0.6px;">CLIENT SATISFACTION - PERCENTAGE CLEANING & IMPUTATION</span>
</div>

```python
# 1. Remove the % sign
df['Client_Satisfaction'] = df['Client_Satisfaction'].astype(str).str.replace('%','')

# 2. Convert 'NaN' text to actual NaN and convert to numeric
df['Client_Satisfaction'] = pd.to_numeric(df['Client_Satisfaction'], errors='coerce')

# 3. Now fill NaN with Median
df['Client_Satisfaction'] = df['Client_Satisfaction'].fillna(df['Client_Satisfaction'].median()).astype(int)
df['Client_Satisfaction']
```

  <span style="font-size:18px;">✨</span>
